```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load dataset and extract first two columns
iris = load_iris()
X = iris.data[:, :2]
y = iris.target

# Split into 70% Training and 30% Testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)

# Define classifiers to evaluate
classifiers = {
    "kNN Classifier (k=3)"   : KNeighborsClassifier(n_neighbors=3),
    "SVM (Linear Kernel)"    : SVC(kernel="linear"),
    "SVM (RBF Kernel)"       : SVC(kernel="rbf"),
    "SVM (Polynomial Kernel)": SVC(kernel="poly")
}

print("--- Classifier Accuracy Metrics ---")
for name, clf in classifiers.items():
    clf.fit(X_train, y_train)
    acc = accuracy_score(y_test, clf.predict(X_test))
    print(f"{name:<25} Accuracy: {acc * 100:.1f}%")
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

iris = load_iris()
X = iris.data[:, :2]

# Fit unsupervised K-Means clustering (using 3 clusters for the 3 species)
kmeans = KMeans(n_clusters=3, init='k-means++', random_state=42)
cluster_labels = kmeans.fit_predict(X)

# Scatter plot generation
plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1], c=cluster_labels, cmap="viridis", edgecolor='k', s=40)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], 
            s=150, c='red', marker='X', label='Centroids')
plt.title("K-Means Flat Clustering Profile (k=3)")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score

# Simulating structural dataset representing Coimbra matrices features (116 records)
np.random.seed(42)
sample_size = 116
mock_data = {
    'Age': np.random.randint(24, 89, size=sample_size),
    'BMI': np.random.uniform(18, 38, size=sample_size),
    'Glucose': np.random.uniform(60, 200, size=sample_size),
    'Insulin': np.random.uniform(2, 50, size=sample_size),
    'HOMA': np.random.uniform(0.5, 25, size=sample_size),
    'Leptin': np.random.uniform(4, 90, size=sample_size),
    'Adiponectin': np.random.uniform(1, 40, size=sample_size),
    'Resistin': np.random.uniform(3, 80, size=sample_size),
    'MCP.1': np.random.uniform(40, 1700, size=sample_size),
    'Classification': np.random.choice([1, 2], size=sample_size) # 1=Healthy, 2=Patient
}
df = pd.DataFrame(mock_data)

# 1. Round off the Leptin feature to two decimal places
df['Leptin'] = df['Leptin'].round(2)

# 2. Extract first 9 attributes (exclude class labels)
X = df.iloc[:, :9]
true_labels = df['Classification']

# 3. Clustering into 2 groups using K-Means
kmeans = KMeans(n_clusters=2, random_state=42)
pred_labels = kmeans.fit_predict(X)

# Map clusters to match true label values (1 and 2), addressing potential label switching
mapped_labels = np.where(pred_labels == 0, 1, 2)
acc_option_1 = accuracy_score(true_labels, mapped_labels)
acc_option_2 = accuracy_score(true_labels, np.where(pred_labels == 0, 2, 1))
final_accuracy = max(acc_option_1, acc_option_2)

print(f"First 3 Rounded Leptin Records:\n{df['Leptin'].head(3).to_list()}")
print(f"Evaluated Clustering Accuracy: {final_accuracy * 100:.2f}%")
```
